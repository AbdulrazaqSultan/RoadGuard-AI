# Detection experiments

Inspect archived validation results. Run training and evaluation from PowerShell using the commands in the README.

```python
import csv
import json

from IPython.display import Image, display

from roadguard.config import load_config
from roadguard.utils.paths import project_root

root = project_root()
config = load_config(root)
experiments = root / config["experiment"]["experiments_root"]
runs = sorted(path.parent for path in experiments.glob("*/metadata.json"))
print("\n".join(run.name for run in runs) if runs else "No training runs are available.")
```

```python
run_name = None  # Replace with a run directory name shown above.
if run_name is not None:
    run = next(path for path in runs if path.name == run_name)
    metadata = json.loads((run / "metadata.json").read_text(encoding="utf-8"))
    display(
        {
            "status": metadata["status"],
            "config": metadata["config"],
            "validation_metrics": metadata.get("validation_metrics"),
        }
    )
    history = run / "results.csv"
    if history.is_file():
        with history.open(encoding="utf-8", newline="") as stream:
            display(list(csv.DictReader(stream))[-5:])
    for plot in ("results.png", "confusion_matrix.png"):
        if (run / plot).is_file():
            display(Image(filename=str(run / plot)))
```
